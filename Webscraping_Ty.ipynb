```python
def get_description(url):
    response = requests.get(url)
    page_soup = BS(response.text)
    all_p = page_soup.find_all('p')
    return all_p[1].text.strip()
```

```python
get_description('https://realpython.github.io/fake-jobs/jobs/television-floor-manager-8.html')
```

```python
jobs_table['Job Description'] = jobs_table['Apply'].apply(get_description)
```
